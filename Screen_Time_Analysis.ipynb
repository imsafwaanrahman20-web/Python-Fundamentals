```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Day":       ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"],
    "Screen_Time_Hours": [3.5, 4.6, 3.8, 6.4, 7.6, 6.0, 5.0],
    "Social_Media":      [2.0, 2.5, 2.1, 3.4, 4.0, 3.2, 2.7],
    "Games":             [0.6, 1.0, 0.7, 1.4, 2.0, 2.1, 1.0],
    "AI_Chatbots":       [0.4, 0.3, 0.7, 1.0, 0.6, 0.4, 0.8],
    "Wikipedia":         [0.3, 0.5, 0.2, 0.4, 0.6, 0.2, 0.3],
    "Newslets":          [0.2, 0.3, 0.1, 0.2, 0.4, 0.1, 0.2]
}

df = pd.DataFrame(data)

# Daily Screen Time

plt.figure(figsize=(10,5))
plt.plot(df["Day"], df["Screen_Time_Hours"], marker="o", linewidth=2, color="royalblue")

plt.title("Daily Screen Time (Hours)")
plt.xlabel("Day of Week")
plt.ylabel("Screen Time (Hours)")
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()

# Daily Screentime Category by App Breakdown
plt.figure(figsize=(12,6))

plt.bar(df["Day"], df["Social_Media"], label="Social Media")
plt.bar(df["Day"], df["Games"], bottom=df["Social_Media"], label="Games")
plt.bar(df["Day"], df["AI_Chatbots"],
        bottom=df["Social_Media"] + df["Games"], label="AI Chatbots")
plt.bar(df["Day"], df["Wikipedia"],
        bottom=df["Social_Media"] + df["Games"] + df["AI_Chatbots"], label="Wikipedia")
plt.bar(df["Day"], df["Newslets"],
        bottom=df["Social_Media"] + df["Games"] + df["AI_Chatbots"] + df["Wikipedia"],
        label="Newslets")

plt.title("Daily Screen Time Breakdown by App Category")
plt.xlabel("Day of Week")
plt.ylabel("Hours")
plt.legend(title="Category")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()

#Total Weekly Screen Time by App category

# Sum each category across the week
totals = df[["Social_Media","Games","AI_Chatbots","Wikipedia","Newslets"]].sum()

# Plot horizontal bar chart
plt.figure(figsize=(10,6))
plt.barh(totals.index, totals.values, color="skyblue")

plt.title("Total Weekly Screen Time by App Category")
plt.xlabel("Total Hours (Weekly)")
plt.ylabel("App Category")
plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.show()

print(df)
```
