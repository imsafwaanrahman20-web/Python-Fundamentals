```python
import pandas as pd
import matplotlib.pyplot as plt


data = {
    "Product_Name": ["Laptop", "Phone", "Tablet", "Mouse", "Headphones", "Printer", "GaN Charger"],
    "Unit_Price": [600, 400, 450, 35, 85, 120, 30],
    "Units_Sold": [120, 200, 150, 500, 350, 90, 600],
    "Total_Revenue": [72000, 80000, 67500, 17500, 29750, 10800, 18000],
    "Date": ["2026-07-01", "2026-07-02", "2026-07-03", "2026-07-04", "2026-07-05", "2026-07-06", "2026-07-07"],
    "Store_Location": ["Starford", "Ilford", "Canary Wharf", "Oxford Street", "Cambridge", "Manchester", "Birmingham"]
}

df = pd.DataFrame(data)

#Revenue Bar Chart.

#Type: Vertical bar chart

#Why: Bar Charts make categorical comparisons

#What it shows: Which product has the highest revenue
plt.figure(figsize=(10, 6))
plt.bar(df["Product_Name"], df["Total_Revenue"], color="blue")
plt.xlabel("Product")
plt.ylabel("Total Revenue (£)")
plt.title("Total Revenue by Product")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



#Price vs Units Sold Scatter Plot

#Why: Reveals relationships between price and demand

#What it shows:
#*   Are expensive items selling well?
#*   Are cheap items selling in huge quantities
#*   Is the top-revenue produce balanced or extreme?

plt.figure(figsize=(10, 6))
plt.scatter(df["Unit_Price"], df["Units_Sold"], color="red")

for i, name in enumerate(df["Product_Name"]):
    plt.annotate(
        name,
        (df["Unit_Price"][i], df["Units_Sold"][i]),
        textcoords="offset points",
        xytext=(10, 10),
        ha="left",
        fontsize=9,
        arrowprops=dict(arrowstyle="->", lw=0.5)
    )

plt.xlabel("Unit Price (£)")
plt.ylabel("Units Sold")
plt.title("Price vs. Units Sold")
plt.grid(True)
plt.tight_layout()
plt.show()
```
